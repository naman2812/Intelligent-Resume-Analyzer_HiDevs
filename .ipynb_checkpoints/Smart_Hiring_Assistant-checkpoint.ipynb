# Smart Hiring Assistant
This notebook demonstrates the Smart Hiring Assistant pipeline, including AI extraction, fuzzy matching A/B tests, and report generation.

```python
from models import Job, Candidate
from parser import parse_resume
from matcher import ab_test_matchers
from reporter import generate_report, export_report_txt
from file_manager import save_to_json, save_to_csv
from ai_integration import AIIntegration

# 1. Setup AI Client (Requires GROQ_API_KEY environment variable)
# Uncomment and set your key if you want to use the AI features:
# import os
# os.environ['GROQ_API_KEY'] = 'gsk_your_api_key_here'

ai_client = AIIntegration()
```

```python
# 2. Define a Job
job = Job(
    title="Senior Python Developer",
    required_skills=["Python", "Django", "REST APIs", "SQL", "AWS"],
    required_experience=5
)
```

```python
# 3. Sample Resume Text
resume_text = """
Name: Alice Johnson
Email: alice@example.com
Experience: 6
Education: B.S. Computer Science
Skills: Python 3, Django Framework, RESTful APIs, MySQL, Docker
"""

# 4. Parse Resume
print("Parsing Resume...")
candidate = parse_resume(resume_text, ai_client=ai_client)
print(f"Candidate Extracted: {candidate.name}, Exp: {candidate.experience}, Skills: {candidate.skills}")
```

```python
# 5. Match Candidate (A/B Test: Strict vs Fuzzy)
print("\n--- A/B Test Matchers ---")
ab_test_result = ab_test_matchers(candidate, job)
strict_result = ab_test_result['strict']
fuzzy_result = ab_test_result['fuzzy']

print(f"Strict Score: {strict_result.score:.2f}%")
print(f"Fuzzy Score:  {fuzzy_result.score:.2f}%")
print(f"Fuzzy matching identified extra skills: {ab_test_result['difference_in_skills']}")
```

```python
# 6. Generate AI Feedback and Final Report
print("\nGenerating AI Feedback...")
feedback = ai_client.generate_recommendation(
    candidate_data=str(candidate),
    job_data=str(job)
)
fuzzy_result.feedback = feedback

report = generate_report(fuzzy_result)
print("\n" + report)
```

```python
# 7. Export Results
print("\nExporting results to files...")
export_report_txt(fuzzy_result, "report.txt")
save_to_json([fuzzy_result], "results.json")
save_to_csv([fuzzy_result], "results.csv")
```
